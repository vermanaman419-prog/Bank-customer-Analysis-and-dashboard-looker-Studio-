# Bank Customer Analysis

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("bank_transactions.csv")
```

```python
df.head()
df.info()
```

STEP 1: Rename Columns

```python
df.columns
```

```python
df = df.rename(columns={
    'TransactionID': 'transaction_id',
    'CustomerID': 'customer_id',
    'CustomerDOB': 'customer_dob',
    'CustGender': 'gender',
    'CustLocation': 'location',
    'CustAccountBalance': 'account_balance',
    'TransactionDate': 'transaction_date',
    'TransactionTime': 'transaction_time',
    'TransactionAmount (INR)': 'transaction_amount'
})
```

STEP 2: Fix Date Columns

```python
df['customer_dob'] = pd.to_datetime(df['customer_dob'], errors='coerce')
df['transaction_date'] = pd.to_datetime(df['transaction_date'], errors='coerce')
```

STEP 3: Create Customer Age & Remove unrealistic ages

```python
df['customer_age'] = (
    (df['transaction_date'] - df['customer_dob'])
    .dt.days // 365
)
```

```python
df.loc[(df['customer_age'] < 18) | (df['customer_age'] > 100), 'customer_age'] = np.nan
```

STEP 4: Clean Gender Column

```python
df['gender'].value_counts(dropna=False)
```

```python
df['gender'] = df['gender'].fillna('Unknown')
df['gender'] = df['gender'].replace({
    'M': 'Male',
    'F': 'Female'
})
```

```python
df['gender'].value_counts(dropna=False)
```

STEP 5: Clean Location Column

```python
df['location'] = df['location'].str.strip()
df['location'] = df['location'].fillna('Unknown')
```

STEP 6: Handle Account Balance

```python
df['account_balance'].isna().sum()
```

```python
median_balance = df['account_balance'].median()
df['account_balance'] = df['account_balance'].fillna(median_balance)
```

STEP 7: Fix Transaction Time

```python
df['transaction_time'] = df['transaction_time'].astype(str).str.zfill(6)
df['transaction_hour'] = df['transaction_time'].str[:2].astype(int)
```

STEP 8: Validate Transaction Amount

```python
df = df[df['transaction_amount'] > 0]
```

STEP 9: Create Business Flags

```python
df['high_value_txn'] = np.where(df['transaction_amount'] > 25000, 'Yes', 'No')
```

```python
df['low_balance_high_txn'] = np.where(
    (df['account_balance'] < 5000) & (df['transaction_amount'] > 20000),
    'Yes',
    'No'
)
```

STEP 10: Final Quality Check

```python
df.info()
df.isna().sum()
df.describe()
```

Step 11 : Further cleaning - Fixing transaction_time, filling null values present in customer_age and standardizing location

```python
# convert to string & remove decimals
df['transaction_time'] = (
    df['transaction_time']
    .astype(str)
    .str.replace('.0', '', regex=False)
    .str.zfill(6)
)

# convert to datetime safely
time_dt = pd.to_datetime(
    df['transaction_time'],
    format='%H%M%S',
    errors='coerce'
)

# replace invalid times with midnight
time_dt = time_dt.fillna(pd.to_datetime('00:00:00'))

# final formatted time
df['transaction_time'] = time_dt.dt.strftime('%H:%M:%S')

# extract hour
df['transaction_hour'] = time_dt.dt.hour
```

```python
# calculate mean age (ignore NaNs automatically)
mean_age = int(round(df['customer_age'].mean()))

# fill only missing values
df['customer_age'] = df['customer_age'].fillna(mean_age)

mean_age
```

```python
# convert to datetime
df['transaction_time_dt'] = pd.to_datetime(
    df['transaction_time'],
    format='%H:%M:%S',
    errors='coerce'
)

# extract hour
df['transaction_hour'] = df['transaction_time_dt'].dt.hour
```

```python
# standard city names (add more if needed)
cities = [
    'MUMBAI', 'PUNE', 'DELHI', 'NEW DELHI', 'AHMEDABAD', 'GURGAON',
    'NOIDA', 'CHENNAI', 'BANGALORE', 'HYDERABAD', 'KOLKATA',
    'LUCKNOW', 'MOHALI', 'THANE', 'ROHTAK', 'GUNTUR', 'JAMSHEDPUR'
]
```

```python
def normalize_city(location):
    if pd.isna(location):
        return location

    loc = location.upper()
    for city in cities:
        if city in loc:
            return city
    return loc  # fallback if no city matched

df['location'] = df['location'].apply(normalize_city)
```

Extraction clean CSV for further analysis and insights through dashboard
